# function with arguments:
example for a function with argument

```python
def addition(a,b):
    return a + b    #it is one of the built-in function in python which is used to add two numbers. and is used in place of "print".
addition(a=24,b=54) #keyword arguments.
     
```

```python
def addition(a,b):
    return a + b 
addition(24,54)  #positional arguments.
     
```

```python
#positional argument
def addition(a,b,c):
    return a + b    
addition(a=24,b=54)
```

```python
#keyword arguments
def addition(a,b,):
    return a + b    
addition(a=24,b=54,c=67)
```

```python
#assaigning multiple values
def addition(*a):
    return a
addition(10,1,23,5,56,8) # multiplication of positional arguments
```

# factorial

```python
def fact(n):  #6
    fact = 1
    for i in range(1,n+1):
        fact *= i
    return fact
fact(n=6)    
```
