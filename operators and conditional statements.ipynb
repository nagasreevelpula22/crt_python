# oprators:
used to perform operations between two operands.
1) arithmetic operators
2) assaignment oprators
3) comparision operators
4) logical operators
5) bitwise operators
6) identity oprators
7) membership operators

```python
#arithmetic operators:
a = 23
b = 64
print("add a+b :", a + b)
print("sub a-b :", a - b)
print("add a*b :", a * b)
print("add a/b :", a / b)
print("add a%b :", a % b)
print("add a//b :", a // b)
print("add a**b :", a**b)
```

```python
#assaignment operators:
a = 10
a +=10
print("add & assign a +=b :", a)
```

```python
#comparision operators:
a = 20
b = 56
print(a>b)
```

```python
# logical operators:
marks = 87
attendance = 67
print(marks >=75 and attendance >=75)
```

```python
# bitwise operators:
print(0 & 1)
print(1 & 1)
print(0 & 0)
print(1 & 0)
```

```python
# identity operators:
a = 24
b = 24
print( a is b)
```

```python
# membership operators:
data = [1, 2, 3, 4, 5,  "ece", "crt"]
crt = "python for ece and cse"
print("ece" in crt)
print("java" in crt)
print(" " in crt)
print("," in data)
print(2 in data)
```

# conditional statement:
if, elif, else

```python
marks = 87
if marks >=75:
    print("first grade")
else:
    print("second grade")    
```

```python
# simple calculator
a= int(input("Enter first number: "))
b= int(input("Enter second number: "))
operator = input("Enter operator: ")

if operator == "+":
    print(a + b)

elif operator == "-":
    print(a - b)

elif operator == "*":
    print(a * b)

elif operator == "/":
    print(a / b)

else:
    print("Invalid operator")
```

```python
# nested conditional statements
n = int(input("Enter a number: "))

if n >= 0:
    if n == 0:
        print("The number is zero")
    else:
        if n % 2 == 0:
            print("Positive even number")
        else:
            print("Positive odd number")
else:
    if n % 2 == 0:
        print("Negative even number")
    else:
        print("Negative odd number")
```

```python
#even or odd
n = int(input("Enter a number: "))

if n % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```
