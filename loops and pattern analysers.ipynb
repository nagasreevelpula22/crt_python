# hollow patterns

```python
n = int(input("Enter n: "))

for i in range(1, n + 1):
    spaces = " " * (n - i)

    if i == 1:
        print(spaces + "*")
    elif i == n:
        print("* " * n)
    else:
        inside = " " * (2 * i - 3)
        print(spaces + "*" + inside + "*")
```

# butterfly star pattern

```python
n = int(input())

# Upper half
for i in range(1, n + 1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)

# Lower half
for i in range(n - 1, 0, -1):
    print("* " * i + "  " * (2 * (n - i)) + "* " * i)
```

# hallow reverse triangle

```python
n = int(input())

for i in range(n, 0, -1):
    if i == n:
        print("* " * i)
    elif i == 1:
        print("*")
    else:
        print("* " + "  " * (i - 2) + "*")
```

# hallow diamond

```python
n= int(input())

for i in range(n, 0, -1):
    spaces = n - i

    if i == n:
        print("* " * i)
    elif i == 1:
        print("  " * spaces + "*")
    else:
        print("  " * spaces + "* " + "  " * (i - 2) + "*")

for i in range(2, n + 1):
    spaces = n - i

    if i == n:
        print("* " * i)
    else:
        print("  " * spaces + "* " + "  " * (i - 2) + "*")
```

# pascals triangle

```python
n= int(input("Enter number of rows: "))

for i in range(n):
    value = 1

    # Spaces for triangle shape
    print(" " * (n - i), end="")

    for j in range(i + 1):
        print(value, end=" ")
        value = value * (i - j) // (j + 1)

    print()
```
