# variables:
 it acts as a container to store the values.
 rules are:
   1) should not use white space
   2) should not use special characters
   3) variables can contain alpha numeric and '_'
   4) variables are case sensitive

```python
ece = "third years"
print(ece)
```

```python
a = 1 
a = 2
a = 6
print(a)
```

# data types
 there will be four built-in data types.
 1) integer (numbers without decimal point either '-' / '+')
 2) float ( numbers with decimal values) 2.456
 3) string ( these are immutable and represented by " " / ' ')
 4) boolean(any two possibilities like True/ False or 0/1)

```python
a = 2
b = 4
c = "ece"
d = True
print(type(d))
```

```python
a = 2
b = 4.52
c = "ece"
d = True
print(type(a))
print(type(b))
print(type(c))
print(type(d))
```
